```python
import pandas as pd 
import numpy as np
```

```python
df=pd.read_csv("Uber.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
# Data Cleaning - Null values .....(Text)

df['Reason for cancelling by Customer'] = df['Reason for cancelling by Customer'].fillna("Not cancelled")
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].fillna("Not cancelled")
df['Incomplete Rides Reason'] = df['Incomplete Rides Reason'].fillna("Not cancelled")
df['Payment Method'] = df['Payment Method'].fillna("NA")
```

```python
# Data Cleaning - Null values .....(Numerical)

df['Driver Ratings'] = df['Driver Ratings'].fillna(df['Driver Ratings'].median())
df['Customer Rating'] = df['Customer Rating'].fillna(df['Customer Rating'].median())
df['Avg VTAT'] = df['Avg VTAT'].fillna(df['Avg VTAT'].median())
df['Avg CTAT'] = df['Avg CTAT'].fillna(df['Avg CTAT'].median())
```

```python
df.isnull().sum()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M:%S').dt.time

print(df['Date'].dtype)
print(df['Time'].dtype)
```

```python
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.strftime('%B')
df['Day_of_Week'] = df['Date'].dt.dayofweek
df['Day_Name'] = df['Date'].dt.strftime('%A')
df['Is_Weekend'] = df['Day_of_Week'].apply(lambda x: 'Weekend' if x >= 5 else 'Weekday')

print(df[['Date', 'Month_Name', 'Day_Name', 'Is_Weekend']].head(10))
```

```python
df['Hour'] = pd.to_datetime(df['Time'].astype(str)).dt.hour
df['Peak_Hour'] = df['Hour'].apply(lambda x: 'Peak' if (8 <= x <= 10) or (17 <= x <= 20) else 'Off-Peak')

print(df[['Time', 'Hour', 'Peak_Hour']].head(10))
```

```python
df.to_csv('uber_final.csv', index=False)
```
